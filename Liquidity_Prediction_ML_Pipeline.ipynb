```
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt

# --- CONFIGURATION ---
TIME_STEPS = 30 # Lookback window for LSTM sequence
EPOCHS = 20
BATCH_SIZE = 64
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# --- 1. Data Simulation (Replacing Data Collection) ---
def create_simulated_data(n_days=730):
    """Simulates 2 years of daily crypto market data (2016-2017)."""
    dates = pd.to_datetime(pd.date_range(start='2016-01-01', periods=n_days, freq='D'))

    # Base Price and Volume are correlated
    price = np.cumsum(np.random.normal(0, 0.1, n_days)) + 50
    volume = np.abs(np.random.normal(50000, 20000, n_days)) + (price * 100)

    # Simulate External Factors (Social Media Activity, Transaction Patterns)
    social_activity = np.sin(np.linspace(0, 2 * np.pi, n_days) * 5) * 10 + 50
    transaction_pattern = np.random.normal(0, 1, n_days) * 5 + 100

    # Target Variable: Liquidity Index (Inversely related to volatility and dependent on Volume/Price)
    volatility = pd.Series(price).pct_change().rolling(window=10).std().fillna(0) * 100
    liquidity_index = (volume / 100000) - (volatility * 0.5) + (social_activity * 0.1)
    liquidity_index = liquidity_index.clip(lower=10).values

    df = pd.DataFrame({
        'Date': dates,
        'Price': price,
        'Volume': volume,
        'Volatility': volatility,
        'SocialActivity': social_activity,
        'TransactionPattern': transaction_pattern,
        'LiquidityIndex': liquidity_index
    })
    df.set_index('Date', inplace=True)
    return df

# --- 2. Data Preprocessing and Feature Engineering ---
def preprocess_and_engineer(df):
    """Handles missing data, normalization, and feature creation."""

    # Handle Missing Values (using simple forward fill for time-series)
    df.fillna(method='ffill', inplace=True)

    # Feature Engineering: Create Lagged Features and Moving Averages
    df['Volume_MA_30'] = df['Volume'].rolling(window=30).mean() # Trend indicator
    df['Price_Lag_7'] = df['Price'].shift(7) # Momentum indicator

    # Drop initial NaNs created by rolling and shifting
    df.dropna(inplace=True)

    # Identify features and target
    features = [col for col in df.columns if col != 'LiquidityIndex']
    target = 'LiquidityIndex'

    # Normalize and scale numerical features
    scaler_X = MinMaxScaler(feature_range=(0, 1))
    X_scaled = scaler_X.fit_transform(df[features])

    scaler_y = MinMaxScaler(feature_range=(0, 1))
    y_scaled = scaler_y.fit_transform(df[[target]])

    return X_scaled, y_scaled, scaler_X, scaler_y, features

# --- 3. Sequence Creation for LSTM ---
def create_sequences(X, y, time_steps):
    """Converts time-series data into sequences for LSTM."""
    Xs, ys = [], []
    for i in range(len(X) - time_steps):
        # Sequence X_t, X_t+1, ..., X_t+time_steps-1
        Xs.append(X[i:(i + time_steps)])
        # Target is Y_t+time_steps
        ys.append(y[i + time_steps])
    return np.array(Xs), np.array(ys)

# --- 4. Model Definition and Training ---
def build_and_train_model(X_train, y_train, input_shape):
    """Builds and trains the LSTM model."""
    print("\n--- Model Selection & Training (LSTM) ---")

    # Deep LSTM architecture for complex sequential patterns
    model = Sequential([
        LSTM(units=100, return_sequences=True, input_shape=input_shape),
        Dropout(0.3), # Preventing overfitting
        LSTM(units=50, return_sequences=False),
        Dropout(0.3), # Preventing overfitting
        Dense(units=1) # Regression output
    ])

    # Compile with Adam optimizer and MSE loss for regression
    model.compile(optimizer='adam', loss='mse')

    print("Model Architecture:")
    model.summary()

    history = model.fit(
        X_train, y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.1,
        verbose=1
    )
    return model, history

# --- 5. Model Evaluation ---
def evaluate_model(model, X_test, y_test, scaler_y):
    """Predicts and evaluates performance using regression metrics."""
    print("\n--- Model Evaluation ---")

    # Predict scaled values
    y_pred_scaled = model.predict(X_test, verbose=0)

    # Inverse transform to get actual liquidity values
    y_test_actual = scaler_y.inverse_transform(y_test)
    y_pred_actual = scaler_y.inverse_transform(y_pred_scaled)

    # Calculate Metrics
    rmse = np.sqrt(mean_squared_error(y_test_actual, y_pred_actual))
    mae = mean_absolute_error(y_test_actual, y_pred_actual)
    r2 = r2_score(y_test_actual, y_pred_actual)

    print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
    print(f"Mean Absolute Error (MAE): {mae:.4f}")
    print(f"R-squared Score (R²): {r2:.4f}")

    return y_test_actual, y_pred_actual

# --- MAIN EXECUTION ---
if __name__ == '__main__':
    # 1. Data Collection (Simulated)
    data = create_simulated_data()
    print("--- 1. Data Collection & Initial EDA ---")
    print(f"Data shape: {data.shape}")

    # 2. Preprocessing & Feature Engineering
    X_scaled, y_scaled, scaler_X, scaler_y, features = preprocess_and_engineer(data)

    # 3. Sequence Creation
    X_sequences, y_targets = create_sequences(X_scaled, y_scaled, TIME_STEPS)

    # 4. Train-Test Split (maintain time order)
    train_size = int(len(X_sequences) * 0.8)
    X_train, X_test = X_sequences[:train_size], X_sequences[train_size:]
    y_train, y_test = y_targets[:train_size], y_targets[train_size:]

    # 5. Build and Train Model
    input_shape = (X_train.shape[1], X_train.shape[2])
    model, history = build_and_train_model(X_train, y_train, input_shape)

    # 6. Save Model and Scaler for Deployment
    model.save('liquidity_lstm_model.h5')

    # Save parameters needed to invert the prediction scaling
    np.save('scaler_y_min.npy', scaler_y.data_min_)
    np.save('scaler_y_max.npy', scaler_y.data_max_)
    np.save('X_test_deployment.npy', X_test)

    print("\nModel artifacts saved: liquidity_lstm_model.h5, scaler_y_min.npy, X_test_deployment.npy")

    # 7. Model Evaluation
    y_test_actual, y_pred_actual = evaluate_model(model, X_test, y_test, scaler_y)

    # Final Visualization
    plt.figure(figsize=(12, 6))
    plt.plot(y_test_actual, label='Actual Liquidity Index')
    plt.plot(y_pred_actual, label='Predicted Liquidity Index')
    plt.title('Cryptocurrency Liquidity Prediction - Test Results')
    plt.xlabel('Time (Test Samples)')
    plt.ylabel('Liquidity Index')
    plt.legend()
    plt.savefig('liquidity_prediction_plot.png')
    print("Evaluation plot saved as liquidity_prediction_plot.png")
```
